```python
# 영상 보간법의 예: 영상 확장 후, 보간법의 차이 보기
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

im = Image.open("/content/drive/MyDrive/Colab Notebooks/week02/clock.jpg")
im_near = im.resize((im.width*5, im.height*5), Image.NEAREST) # 영순위 보간, 가장 가까운 이웃
im_linear = im.resize((im.width*5, im.height*5), Image.BILINEAR) # 양선형보간
im_cubic = im.resize((im.width*5, im.height*5), Image.BICUBIC) # 입방보간

plt.figure(figsize=(10,10))
#plt.figure(figsize=(30,30)) # 화면 픽셀 크기를 크게 보이기
plt.subplot(221), plt.imshow(im), plt.axis('off')
plt.subplot(222), plt.imshow(im_near), plt.axis('off')
plt.subplot(223), plt.imshow(im_linear), plt.axis('off')
plt.subplot(224), plt.imshow(im_cubic), plt.axis('off')
plt.show()
```

```python
# 영상 기하학적 변환
im = Image.open("/content/drive/MyDrive/Colab Notebooks/week02/parrot.png")
im_crop = im.crop((175,75,320,200)) # 자르기(crop)
im_rotate = im.rotate(45) # 반시계 방향 회전
im_reflect = im.transpose(Image.FLIP_LEFT_RIGHT) # 좌우 반전, 180도 회전
im_shear = im.transform((int(1.4*im.width), im.height), Image.AFFINE, data=(1,-0.5,0,0,1,0))
# 어파인 변환, x축으로 밀기, shearing

plt.figure(figsize=(20, 10))
plt.subplot(231), plt.imshow(im), plt.axis('off')
plt.subplot(232), plt.imshow(im_crop), plt.axis('off')
plt.subplot(233), plt.imshow(im_rotate), plt.axis('off')
plt.subplot(234), plt.imshow(im_reflect), plt.axis('off')
plt.subplot(235), plt.imshow(im_shear), plt.axis('off')
#plt.subplots_adjust(wspace=0.0, hspace=0.01) # 여러 영상 표시시 공간 조정
plt.show()
```

```python
# 가중치를 활용한 디지털 합성
import numpy as np
import matplotlib.pylab as plt
import matplotlib.image as mpimg # matplot 내에 영상 입출력 라이브러리

im1 = mpimg.imread("/content/drive/MyDrive/Colab Notebooks/week02/messi.jpg") / 255
im2 = mpimg.imread("/content/drive/MyDrive/Colab Notebooks/week02/ronaldo.jpg") / 255

i = 1
plt.figure(figsize=(15,15))
for alpha in np.linspace(0,1,20): # 0.0, 0.05, 0.1, 0.15, .... 1.0 (20개 반위)
# 구문안에 띄어 쓰기를 정확히 맞추어 블럭 범위 설정
    plt.subplot(4,5,i)
    plt.imshow((1-alpha)*im1 + alpha*im2) # 가중치에 의한 영상 합성
    plt.axis('off')
    i += 1
plt.subplots_adjust(wspace=0.05, hspace=0.01)
plt.show()
```
